## The negative of a vector

In session 1's zero vector section, the second move undid the first one. Here's that idea again: the animation draws a vector $\mathbf{a}$, then the move that takes you straight back, which is called $-\mathbf{a}$.

**Before you run it:** if $\mathbf{a} = \begin{pmatrix} 3 \\ 2 \end{pmatrix}$, what column vector takes you straight back?

```python
%%manim -ql -v WARNING --progress_bar none FlipIt

# ✏️ Change these
v = (3, 2)
start = (-2, -1)


def describe(x, y):
    # (3, -2) -> "3 right, 2 down"
    parts = []
    if x != 0:
        parts.append(f"{abs(x)} {'right' if x > 0 else 'left'}")
    if y != 0:
        parts.append(f"{abs(y)} {'up' if y > 0 else 'down'}")
    return ", ".join(parts)


class FlipIt(Scene):
    def construct(self):
        plane = NumberPlane()
        tail = plane.c2p(*start)
        tip = plane.c2p(start[0] + v[0], start[1] + v[1])
        self.add(plane, Dot(tail))

        forward = Arrow(tail, tip, buff=0, color=YELLOW)
        back = Arrow(tip, tail, buff=0, color=PINK)
        a_label = Text(f"a: {describe(*v)}", font_size=28, color=YELLOW)
        minus_a_label = Text(f"−a: {describe(-v[0], -v[1])}", font_size=28, color=PINK)
        labels = VGroup(a_label, minus_a_label).arrange(DOWN, aligned_edge=LEFT)
        labels.to_corner(UL).add_background_rectangle(opacity=0.85, buff=0.2)

        self.play(GrowArrow(forward), FadeIn(labels.background_rectangle), Write(a_label))
        self.play(GrowArrow(back), Write(minus_a_label))
        self.wait(0.5)

        # Slide −a so that both arrows start from the same point
        self.play(back.animate.shift(tail - tip), run_time=1.5)
        self.wait()
```

> ### Definition: the negative of a vector
>
> The **negative** of $\mathbf{a}$, written $-\mathbf{a}$, has the **same length** as $\mathbf{a}$ but points in the **opposite direction**. To find it, flip the sign of both numbers:
>
> $$-\begin{pmatrix} 3 \\ 2 \end{pmatrix} = \begin{pmatrix} -3 \\ -2 \end{pmatrix}$$
>
> - Doing $\mathbf{a}$ and then $-\mathbf{a}$ gets you back where you started, so the total movement is $\mathbf{0}$.
> - Swapping the letters reverses a vector: $\overrightarrow{BA} = -\overrightarrow{AB}$.

## Parallelograms

A **parallelogram** is a four-sided shape whose opposite sides are **parallel** and the **same length**. Rectangles, squares and rhombuses are all special kinds of parallelogram.

The corners are labelled **in order as you go round the shape**. So in parallelogram $ABCD$ you go $A \to B \to C \to D$ and back to $A$. $AB$ and $DC$ are opposite sides, and $AC$ is a diagonal, not a side.

Exam questions usually give you two of the sides as vectors, like this:

```python
%%manim -ql -v WARNING --progress_bar none ParallelogramABCD

# ✏️ Change these
A = (-3, -2)
a = (5, 0)  # the move from A to B
b = (2, 3)  # the move from A to D


class ParallelogramABCD(Scene):
    def construct(self):
        plane = NumberPlane()
        self.add(plane)

        B = (A[0] + a[0], A[1] + a[1])
        D = (A[0] + b[0], A[1] + b[1])
        C = (B[0] + b[0], B[1] + b[1])
        corners = [plane.c2p(*P) for P in (A, B, C, D)]
        centre = sum(corners) / 4

        def outside(point, distance=0.4):
            # A spot just outside the shape, on the far side from the centre
            direction = point - centre
            return point + distance * direction / np.linalg.norm(direction)

        self.play(Create(Polygon(*corners, color=WHITE, fill_opacity=0.15)))
        for name, point in zip("ABCD", corners):
            self.play(Write(Text(name, font_size=32).move_to(outside(point))), run_time=0.4)

        ab = Arrow(corners[0], corners[1], buff=0, color=YELLOW)
        ad = Arrow(corners[0], corners[3], buff=0, color=PINK)
        a_label = Text("a", font_size=32, weight=BOLD, color=YELLOW).move_to(outside(ab.get_center()))
        b_label = Text("b", font_size=32, weight=BOLD, color=PINK).move_to(outside(ad.get_center()))
        self.play(GrowArrow(ab), Write(a_label))
        self.play(GrowArrow(ad), Write(b_label))
        self.wait()
```

## Adding vectors

Adding two vectors means **doing one move, then the other**. You draw the second arrow starting from the tip of the first one: this is called **tip to tail**.

The animation draws $\mathbf{a}$ in green, then $\mathbf{b}$ in red starting from the tip of $\mathbf{a}$. Then it draws the total movement $\mathbf{a} + \mathbf{b}$ in yellow and works it out as a column vector.

**Before you run it:** with $\mathbf{a} = \begin{pmatrix} 3 \\ 1 \end{pmatrix}$ and $\mathbf{b} = \begin{pmatrix} 1 \\ 2 \end{pmatrix}$, where does the yellow arrow end? Write $\mathbf{a} + \mathbf{b}$ as a column vector.

```python
%%manim -ql -v WARNING --progress_bar none TipToTail

# ✏️ Change these
a = (3, 1)
b = (1, 2)


def column(x, y, color):
    # A column vector drawn with text, so no LaTeX is needed
    top = Text(str(x).replace("-", "−"), font_size=34, color=color)
    bottom = Text(str(y).replace("-", "−"), font_size=34, color=color)
    entries = VGroup(top, bottom).arrange(DOWN, buff=0.2)
    left = Text("(", font_size=34).stretch_to_fit_height(entries.height * 1.3).next_to(entries, LEFT, buff=0.12)
    right = Text(")", font_size=34).stretch_to_fit_height(entries.height * 1.3).next_to(entries, RIGHT, buff=0.12)
    return VGroup(left, entries, right)


def arrow(start, end, color):
    # A zero-length arrow can't be drawn, so use a dot instead
    if np.allclose(start, end):
        return Dot(start, color=color)
    return Arrow(start, end, buff=0, color=color)


def grow(mob):
    return GrowArrow(mob) if isinstance(mob, Arrow) else FadeIn(mob)


def emptiest_corner(points):
    # The screen corner furthest from all the arrow ends, so the working doesn't cover them
    corners = [UL, UR, DL, DR]
    return max(corners, key=lambda c: min(np.linalg.norm(c * np.array([7, 4, 0]) - p) for p in points))


class TipToTail(Scene):
    def construct(self):
        plane = NumberPlane()
        self.add(plane, Dot(plane.c2p(0, 0)))
        total = (a[0] + b[0], a[1] + b[1])
        O, A, B, T = plane.c2p(0, 0), plane.c2p(*a), plane.c2p(*b), plane.c2p(*total)

        arrow_a = arrow(O, A, GREEN)
        arrow_b = arrow(A, T, RED)
        arrow_total = arrow(O, T, YELLOW)

        col_a, col_b, col_total = column(*a, GREEN), column(*b, RED), column(*total, YELLOW)
        plus, equals = Text("+", font_size=34), Text("=", font_size=34)
        working = VGroup(col_a, plus, col_b, equals, col_total).arrange(RIGHT, buff=0.25)
        message = Text("b first, then a: same place", font_size=26)
        panel = VGroup(working, message).arrange(DOWN, buff=0.3)
        panel.to_corner(emptiest_corner([O, A, B, T])).add_background_rectangle(opacity=0.85, buff=0.2)

        # a, then b from the tip of a, then the total
        self.play(grow(arrow_a), FadeIn(panel.background_rectangle), FadeIn(col_a))
        self.play(grow(arrow_b), FadeIn(plus), FadeIn(col_b))
        self.play(grow(arrow_total))

        # Add the tops, then add the bottoms
        self.play(FadeIn(equals), FadeIn(col_total[0]), FadeIn(col_total[2]))
        self.play(TransformFromCopy(VGroup(col_a[1][0], col_b[1][0]), col_total[1][0]))
        self.play(TransformFromCopy(VGroup(col_a[1][1], col_b[1][1]), col_total[1][1]))
        self.wait(0.5)

        # The other order: b first, then a
        b_first = arrow(O, B, RED).set_opacity(0.5)
        a_second = arrow(B, T, GREEN).set_opacity(0.5)
        self.play(grow(b_first))
        self.play(grow(a_second))
        self.play(Write(message))
        self.wait()
```

> ### Definition: adding vectors
>
> $\mathbf{a} + \mathbf{b}$ means "do $\mathbf{a}$, then do $\mathbf{b}$". On a diagram, draw $\mathbf{b}$ starting from the tip of $\mathbf{a}$ (**tip to tail**). The total movement goes from the start of $\mathbf{a}$ to the tip of $\mathbf{b}$.
>
> With column vectors, **add the tops and add the bottoms**:
>
> $$\begin{pmatrix} 3 \\ 1 \end{pmatrix} + \begin{pmatrix} 1 \\ 2 \end{pmatrix} = \begin{pmatrix} 3 + 1 \\ 1 + 2 \end{pmatrix} = \begin{pmatrix} 4 \\ 3 \end{pmatrix}$$
>
> - **The order doesn't matter:** $\mathbf{a} + \mathbf{b} = \mathbf{b} + \mathbf{a}$. The two orders are the two routes round a parallelogram.
> - The "across then up" animation from session 1 was secretly an addition: $\begin{pmatrix} 3 \\ 0 \end{pmatrix} + \begin{pmatrix} 0 \\ 2 \end{pmatrix} = \begin{pmatrix} 3 \\ 2 \end{pmatrix}$.

## Multiplying a vector by a number

In the last question you worked out $\mathbf{a} + \mathbf{a} + \mathbf{a}$. The shorter way to write it is $3\mathbf{a}$. Multiplying a vector by a number like this is called **scalar multiplication**, and the number is called a **scalar**.

The animation draws $\mathbf{a}$ in yellow, then smoothly changes the number $k$ from $1$ to whatever you choose, showing $k\mathbf{a}$ in orange as it goes.

**Before you run it:** with $\mathbf{a} = \begin{pmatrix} 3 \\ 1 \end{pmatrix}$ and $k = 2$, write down $2\mathbf{a}$ as a column vector. What will the orange arrow look like compared to the yellow one?

```python
%%manim -ql -v WARNING --progress_bar none Stretch

# ✏️ Change these
a = (3, 1)
k = 2


def fmt(n):
    # 2.0 -> "2", 1.5 -> "1.5", -0.333 -> "−0.33"
    return f"{round(n, 2):g}".replace("-", "−")


def column(x, y, color):
    # A column vector drawn with text, so no LaTeX is needed
    top = Text(fmt(x), font_size=34, color=color)
    bottom = Text(fmt(y), font_size=34, color=color)
    entries = VGroup(top, bottom).arrange(DOWN, buff=0.2)
    left = Text("(", font_size=34).stretch_to_fit_height(entries.height * 1.3).next_to(entries, LEFT, buff=0.12)
    right = Text(")", font_size=34).stretch_to_fit_height(entries.height * 1.3).next_to(entries, RIGHT, buff=0.12)
    return VGroup(left, entries, right)


def arrow(start, end, color):
    # A zero-length arrow can't be drawn, so use a dot instead
    if np.linalg.norm(end - start) < 0.05:
        return Dot(start, color=color)
    return Arrow(start, end, buff=0, color=color)


def emptiest_corner(points):
    # The screen corner furthest from all the arrow ends, so the working doesn't cover them
    corners = [UL, UR, DL, DR]
    return max(corners, key=lambda c: min(np.linalg.norm(c * np.array([7, 4, 0]) - p) for p in points))


class Stretch(Scene):
    def construct(self):
        plane = NumberPlane()
        O, A, KA = plane.c2p(0, 0), plane.c2p(*a), plane.c2p(k * a[0], k * a[1])
        self.add(plane, Dot(O))

        # The line through a: keep an eye on it
        if a != (0, 0):
            direction = (A - O) / np.linalg.norm(A - O)
            self.add(DashedLine(O - 20 * direction, O + 20 * direction, color=GREY_A, stroke_opacity=0.9))

        # Working: k × a = ka, with a slot at the top for the live value of k
        k_word = Text(f"{fmt(k)} ×", font_size=34, color=ORANGE)
        col_a, col_ka = column(*a, YELLOW), column(k * a[0], k * a[1], ORANGE)
        working = VGroup(k_word, col_a, Text("=", font_size=34), col_ka).arrange(RIGHT, buff=0.25)
        k_slot = Text("k = −0.00", font_size=30)
        panel = VGroup(k_slot, working).arrange(DOWN, buff=0.3)
        panel.to_corner(emptiest_corner([O, A, KA])).add_background_rectangle(opacity=0.85, buff=0.2)

        original = arrow(O, A, YELLOW)
        self.play(GrowArrow(original) if isinstance(original, Arrow) else FadeIn(original),
                  FadeIn(panel.background_rectangle))
        self.add(original.copy().set_opacity(0.5))  # a faint copy of a stays behind

        # Slide k from 1 to the chosen value
        tracker = ValueTracker(1)
        moving = always_redraw(lambda: arrow(O, plane.c2p(tracker.get_value() * a[0],
                                                            tracker.get_value() * a[1]), ORANGE))
        k_live = always_redraw(lambda: Text(f"k = {tracker.get_value():.2f}".replace("-0.00", "0.00").replace("-", "−"),
                                            font_size=30).move_to(k_slot))
        self.remove(original)
        self.add(moving, k_live)
        self.play(tracker.animate.set_value(k), run_time=3)

        # Multiply the top by k, then the bottom by k
        self.play(FadeIn(k_word), FadeIn(col_a), FadeIn(working[2]), FadeIn(col_ka[0]), FadeIn(col_ka[2]))
        self.play(TransformFromCopy(VGroup(k_word, col_a[1][0]), col_ka[1][0]))
        self.play(TransformFromCopy(VGroup(k_word, col_a[1][1]), col_ka[1][1]))
        self.wait()
```

> ### Definition: scalar multiplication
>
> To multiply a vector by a number $k$, **multiply both numbers in the column vector by $k$**:
>
> $$3\begin{pmatrix} 2 \\ -1 \end{pmatrix} = \begin{pmatrix} 3 \times 2 \\ 3 \times (-1) \end{pmatrix} = \begin{pmatrix} 6 \\ -3 \end{pmatrix}$$
>
> - If $k$ is more than $1$, the arrow **stretches**. If $k$ is between $0$ and $1$, it **shrinks**.
> - If $k$ is negative, the arrow also **flips round**. $k = -1$ gives $-\mathbf{a}$.
> - If $k = 0$, you get the zero vector $\mathbf{0}$.
> - $3\mathbf{a}$ is the same as $\mathbf{a} + \mathbf{a} + \mathbf{a}$.
>
> ### Parallel vectors
>
> Every multiple of $\mathbf{a}$ lies along the same line as $\mathbf{a}$. So:
>
> **Two vectors are parallel when one is a multiple of the other.**
>
> For example, $\begin{pmatrix} 6 \\ -3 \end{pmatrix}$ is parallel to $\begin{pmatrix} 2 \\ -1 \end{pmatrix}$ because it is $3$ times it. This is the main tool in the hardest GCSE vector questions.

## Subtracting vectors

With ordinary numbers, taking away $2$ is the same as adding $-2$: $\;5 - 2 = 5 + (-2)$. Vectors work the same way.

The animation draws $\mathbf{a}$ in green and $\mathbf{b}$ in red from the tip of $\mathbf{a}$. Then it **spins $\mathbf{b}$ round** to make $-\mathbf{b}$, and draws the total movement $\mathbf{a} - \mathbf{b}$ in yellow. After that, it shows you a second way to see $\mathbf{a} - \mathbf{b}$.

**Before you run it:** with $\mathbf{a} = \begin{pmatrix} 3 \\ 1 \end{pmatrix}$ and $\mathbf{b} = \begin{pmatrix} 1 \\ 2 \end{pmatrix}$, write down $\mathbf{a} - \mathbf{b}$ as a column vector.

```python
%%manim -ql -v WARNING --progress_bar none TakeAway

# ✏️ Change these
a = (3, 1)
b = (1, 2)


def column(x, y, color):
    # A column vector drawn with text, so no LaTeX is needed
    top = Text(str(x).replace("-", "−"), font_size=34, color=color)
    bottom = Text(str(y).replace("-", "−"), font_size=34, color=color)
    entries = VGroup(top, bottom).arrange(DOWN, buff=0.2)
    left = Text("(", font_size=34).stretch_to_fit_height(entries.height * 1.3).next_to(entries, LEFT, buff=0.12)
    right = Text(")", font_size=34).stretch_to_fit_height(entries.height * 1.3).next_to(entries, RIGHT, buff=0.12)
    return VGroup(left, entries, right)


def arrow(start, end, color):
    # A zero-length arrow can't be drawn, so use a dot instead
    if np.allclose(start, end):
        return Dot(start, color=color)
    return Arrow(start, end, buff=0, color=color)


def grow(mob):
    return GrowArrow(mob) if isinstance(mob, Arrow) else FadeIn(mob)


def emptiest_corner(points):
    # The screen corner furthest from all the arrow ends, so the working doesn't cover them
    corners = [UL, UR, DL, DR]
    return max(corners, key=lambda c: min(np.linalg.norm(c * np.array([7, 4, 0]) - p) for p in points))


class TakeAway(Scene):
    def construct(self):
        plane = NumberPlane()
        self.add(plane, Dot(plane.c2p(0, 0)))
        diff = (a[0] - b[0], a[1] - b[1])
        O, A, B, T = plane.c2p(0, 0), plane.c2p(*a), plane.c2p(*b), plane.c2p(*diff)

        col_a, col_b, col_diff = column(*a, GREEN), column(*b, RED), column(*diff, YELLOW)
        minus, equals = Text("−", font_size=34), Text("=", font_size=34)
        working = VGroup(col_a, minus, col_b, equals, col_diff).arrange(RIGHT, buff=0.25)
        # Size the panel for the longer of the two messages
        message = Text("taking away b = adding −b", font_size=26)
        new_message = Text("a − b goes from the tip of b to the tip of a", font_size=26)
        panel = VGroup(working, new_message).arrange(DOWN, buff=0.3)
        panel.to_corner(emptiest_corner([O, A, B, T, A + (B - O)]))
        panel.add_background_rectangle(opacity=0.85, buff=0.2)
        message.move_to(new_message)

        # Part 1: a, then b from the tip of a, then spin b round to make −b
        arrow_a = arrow(O, A, GREEN)
        arrow_b = arrow(A, A + (B - O), RED)
        self.play(grow(arrow_a), FadeIn(panel.background_rectangle), FadeIn(col_a))
        self.play(grow(arrow_b), FadeIn(minus), FadeIn(col_b))
        self.play(Rotate(arrow_b, PI, about_point=A), arrow_b.animate.set_color(PINK), Write(message), run_time=1.5)

        arrow_diff = arrow(O, T, YELLOW)
        self.play(grow(arrow_diff))

        # Subtract the tops, then subtract the bottoms
        self.play(FadeIn(equals), FadeIn(col_diff[0]), FadeIn(col_diff[2]))
        self.play(TransformFromCopy(VGroup(col_a[1][0], col_b[1][0]), col_diff[1][0]))
        self.play(TransformFromCopy(VGroup(col_a[1][1], col_b[1][1]), col_diff[1][1]))
        self.wait()

        # Part 2: a and b from the same point. a − b fills the gap between their tips
        self.play(FadeOut(arrow_b))
        self.play(grow(arrow(O, B, RED)))
        self.play(arrow_diff.copy().animate.shift(B - O), Transform(message, new_message), run_time=1.5)
        self.wait()
```

> ### Definition: subtracting vectors
>
> $\mathbf{a} - \mathbf{b}$ means $\mathbf{a} + (-\mathbf{b})$: do $\mathbf{a}$, then do $\mathbf{b}$ **backwards**.
>
> With column vectors, **subtract the tops and subtract the bottoms**:
>
> $$\begin{pmatrix} 3 \\ 1 \end{pmatrix} - \begin{pmatrix} 1 \\ 2 \end{pmatrix} = \begin{pmatrix} 3 - 1 \\ 1 - 2 \end{pmatrix} = \begin{pmatrix} 2 \\ -1 \end{pmatrix}$$
>
> - **The order matters:** $\mathbf{b} - \mathbf{a} = -(\mathbf{a} - \mathbf{b})$. Same length, opposite direction.
> - If $\mathbf{a}$ and $\mathbf{b}$ start from the same point, then $\mathbf{a} - \mathbf{b}$ is the arrow **from the tip of $\mathbf{b}$ to the tip of $\mathbf{a}$**. You'll use this all the time in the next session.
